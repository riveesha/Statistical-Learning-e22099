```python
# Install required package (Run this once in Google Colab)
!pip install plotly scipy

from IPython.display import display, Math, Markdown
import numpy as np
import plotly.graph_objects as go
from scipy.stats import beta

# ---------------------------------------------------
# TEXT DISPLAY SECTION
# ---------------------------------------------------

display(Markdown("# Beta Distribution Solution"))

# Step 1
display(Markdown("## 1. Probability Space"))

display(Math(r"\Omega=[0,1], \qquad \mathfrak{F}=\mathfrak{B}([0,1])"))

display(Markdown(
"where $\\mathfrak{B}([0,1])$ is the Borel $\\sigma$-algebra on $[0,1]$."
))

display(Math(r"P(A)=\int_A f(x)\,dx, \qquad A\in\mathfrak{F}"))

display(Math(r"""
f(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1,\\
0, & \text{otherwise}.
\end{cases}
"""))

# Step 2
display(Markdown("## 2. Random Variable"))

display(Math(r"X(\omega)=\omega, \qquad \omega\in[0,1]"))

display(Markdown(
"Thus $X$ records the defective proportion in the batch."
))

# Step 3
display(Markdown("## 3. Probability Density Function"))

display(Math(r"""
f_X(x)=
\begin{cases}
\dfrac{1}{B(2,18)}x(1-x)^{17}, & 0<x<1,\\
0, & \text{otherwise}.
\end{cases}
"""))

# Step 4
display(Markdown("## 4. Verification of Valid Density"))

display(Math(r"f(x)\ge0 \quad \text{for all } x"))

display(Math(r"""
\int_{-\infty}^{\infty} f(x)\,dx
=
\int_0^1 \dfrac{1}{B(2,18)}x(1-x)^{17}\,dx
"""))

display(Math(r"""
=
\dfrac{1}{B(2,18)}B(2,18)
=1
"""))

display(Markdown("So $f$ is a valid probability density function."))

# Step 5
display(Markdown("## 5. Expectation"))

display(Math(r"E[X]=\frac{\alpha}{\alpha+\beta}=\frac{2}{20}=0.10"))

# Step 6
display(Markdown("## 6. Variance"))

display(Math(r"""
\mathrm{Var}(X)=
\frac{\alpha\beta}
{(\alpha+\beta)^2(\alpha+\beta+1)}
"""))

display(Math(r"""
\mathrm{Var}(X)=
\frac{2\cdot18}{20^2\cdot21}
=
\frac{36}{8400}
=
\frac{3}{700}
\approx0.00429
"""))

# Step 7
display(Markdown("## 7. Probability that $X \\le 0.10$"))

display(Math(r"""
P(X\le0.10)
=
\int_0^{0.10}
\dfrac{1}{B(2,18)}x(1-x)^{17}\,dx
"""))

display(Math(r"""
P(X\le0.10)=I_{0.10}(2,18)
"""))

# Step 8
display(Markdown("## 8. Probability that $0.05 \\le X \\le 0.15$"))

display(Math(r"""
P(0.05\le X\le0.15)
=
\int_{0.05}^{0.15}
\dfrac{1}{B(2,18)}x(1-x)^{17}\,dx
"""))

display(Math(r"""
P(0.05\le X\le0.15)
=
I_{0.15}(2,18)-I_{0.05}(2,18)
"""))

# Step 9
display(Markdown("## 9. Why Beta Distribution is Suitable"))

display(Markdown(
"The Beta distribution is appropriate because the defective proportion is a continuous value between 0 and 1, and Beta distributions are designed to model proportions and probabilities."
))

# Step 10
display(Markdown("## 10. Interpretation"))

display(Markdown(
"The expectation **E[X]=0.10** means that the long-run average defective proportion is **10%**."
))

display(Markdown(
"The variance **Var(X)≈0.00429** measures how much the defective proportion varies from batch to batch."
))

# ---------------------------------------------------
# GRAPH SECTION
# ---------------------------------------------------

display(Markdown("# Beta Distribution Graph"))

def plot_beta_distribution(alpha, beta_val):

    # x values
    x = np.linspace(0, 1, 1000)

    # Beta PDF values
    y = beta.pdf(x, alpha, beta_val)

    # Create figure
    fig = go.Figure()

    fig.add_trace(go.Scatter(
        x=x,
        y=y,
        mode='lines',
        name=f'Beta(alpha={alpha}, beta={beta_val})',
        line=dict(width=3)
    ))

    # Layout
    fig.update_layout(
        title=f'Beta Distribution PDF (alpha={alpha}, beta={beta_val})',
        xaxis_title='x',
        yaxis_title='f(x)',
        template='plotly_white'
    )

    return fig

# Display graph at the bottom
fig = plot_beta_distribution(alpha=2, beta_val=18)
fig.show()
```

```python
# Install required packages (Run once in Google Colab)
!pip install plotly scipy

from IPython.display import display, Math, Markdown
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang

# ---------------------------------------------------
# TEXT DISPLAY SECTION
# ---------------------------------------------------

display(Markdown("# Triangular Distribution Solution"))

# Step 1
display(Markdown("## 1. Support of the Random Variable"))

display(Math(r"4 \leq X \leq 10"))

# Step 2
display(Markdown("## 2. Probability Density Function"))

display(Math(r"""
f_X(x)=
\begin{cases}
\dfrac{2(x-4)}{(10-4)(6-4)}, & 4 \leq x < 6,\\
\dfrac{2}{10-4}, & x=6,\\
\dfrac{2(10-x)}{(10-4)(10-6)}, & 6 < x \leq 10,\\
0, & \text{otherwise}.
\end{cases}
"""))

display(Markdown("This simplifies to"))

display(Math(r"""
f_X(x)=
\begin{cases}
\dfrac{x-4}{6}, & 4 \leq x < 6,\\
\dfrac{1}{3}, & x=6,\\
\dfrac{10-x}{12}, & 6 < x \leq 10,\\
0, & \text{otherwise}.
\end{cases}
"""))

# Step 3
display(Markdown("## 3. Verification that $f_X$ is a Valid Density"))

display(Math(r"""
\int_{4}^{6} \frac{x-4}{6}\,dx
+
\int_{6}^{10} \frac{10-x}{12}\,dx
"""))

display(Markdown("For the first part:"))

display(Math(r"""
\int_{4}^{6} \frac{x-4}{6}\,dx
=
\frac{1}{6}\int_{4}^{6}(x-4)\,dx
=
\frac{1}{6}\cdot2
=
\frac{1}{3}
"""))

display(Markdown("For the second part:"))

display(Math(r"""
\int_{6}^{10} \frac{10-x}{12}\,dx
=
\frac{1}{12}\int_{6}^{10}(10-x)\,dx
=
\frac{1}{12}\cdot8
=
\frac{2}{3}
"""))

display(Math(r"""
\int_{-\infty}^{\infty} f_X(x)\,dx
=
\frac{1}{3}+\frac{2}{3}
=1
"""))

display(Markdown("So $f_X$ is a valid probability density function."))

# Step 4
display(Markdown("## 4. Expectation"))

display(Math(r"""
E[X]
=
\frac{a+b+c}{3}
"""))

display(Math(r"""
E[X]
=
\frac{4+10+6}{3}
=
\frac{20}{3}
\approx6.67
"""))

# Step 5
display(Markdown("## 5. Variance"))

display(Math(r"""
\mathrm{Var}(X)
=
\frac{
a^2+b^2+c^2-ab-ac-bc
}{18}
"""))

display(Math(r"""
\mathrm{Var}(X)
=
\frac{
4^2+10^2+6^2-(4)(10)-(4)(6)-(10)(6)
}{18}
"""))

display(Math(r"""
=
\frac{16+100+36-40-24-60}{18}
=
\frac{28}{18}
=
\frac{14}{9}
\approx1.56
"""))

# Step 6
display(Markdown("## 6. Interpretation of Parameters"))

display(Markdown("""
- $a=4$ is the shortest possible installation time
- $b=10$ is the longest possible installation time
- $c=6$ is the most likely installation time
"""))

# Step 7
display(Markdown("## 7. Why Triangular Distribution is Suitable"))

display(Markdown(
"The Triangular distribution is useful when only the minimum, maximum, and most likely estimates are known. It is commonly used in project planning and estimation problems."
))

# Step 8
display(Markdown("## 8. Computing $P(X \\leq 5)$"))

display(Math(r"""
P(X\leq5)
=
\int_{4}^{5}\frac{x-4}{6}\,dx
"""))

display(Math(r"""
=
\frac{1}{6}\int_{4}^{5}(x-4)\,dx
=
\frac{1}{6}\cdot\frac{1}{2}
=
\frac{1}{12}
"""))

display(Math(r"""
P(X\leq5)
=
\frac{1}{12}
\approx0.0833
"""))

# Step 9
display(Markdown("## 9. Computing $P(5 \\leq X \\leq 8)$"))

display(Math(r"""
P(5\leq X\leq8)
=
\int_{5}^{6}\frac{x-4}{6}\,dx
+
\int_{6}^{8}\frac{10-x}{12}\,dx
"""))

display(Markdown("First part:"))

display(Math(r"""
\int_{5}^{6}\frac{x-4}{6}\,dx
=
\frac{1}{6}\int_{5}^{6}(x-4)\,dx
=
\frac{1}{6}\cdot\frac{3}{2}
=
\frac{1}{4}
"""))

display(Markdown("Second part:"))

display(Math(r"""
\int_{6}^{8}\frac{10-x}{12}\,dx
=
\frac{1}{12}\int_{6}^{8}(10-x)\,dx
=
\frac{1}{12}\cdot6
=
\frac{1}{2}
"""))

display(Math(r"""
P(5\leq X\leq8)
=
\frac{1}{4}+\frac{1}{2}
=
\frac{3}{4}
"""))

# Step 10
display(Markdown("## 10. Practical Use"))

display(Markdown(
"This model helps estimate project completion times and analyze uncertainty in planning, scheduling, staffing, and contingency management."
))

# ---------------------------------------------------
# GRAPH SECTION
# ---------------------------------------------------

display(Markdown("# Triangular Distribution Graph"))

# Parameters
a = 4
b = 10
c_mode = 6

# Scipy parameterization
loc = a
scale = b - a
c_scipy = (c_mode - a) / (b - a)

# Generate points
x = np.linspace(0, b + 1, 500)
y = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)

# Create figure
fig = go.Figure()

fig.add_trace(go.Scatter(
    x=x,
    y=y,
    mode='lines',
    fill='tozeroy',
    name='Triangular Distribution',
    line=dict(width=3)
))

# Layout
fig.update_layout(
    title=f'Triangular Distribution (a={a}, b={b}, c={c_mode})',
    xaxis_title='Installation Time (hours)',
    yaxis_title='Probability Density',
    template='plotly_white'
)

# Show graph
fig.show()
```

```python
# Install required packages (Run once in Google Colab)
!pip install plotly scipy

from IPython.display import display, Math, Markdown
import numpy as np
import plotly.graph_objects as go
from scipy.stats import triang, beta

# ---------------------------------------------------
# TEXT DISPLAY SECTION
# ---------------------------------------------------

display(Markdown("# PERT Distribution Solution"))

# Step 1
display(Markdown("## 1. PERT Shape Parameters"))

display(Math(r"a=8, \qquad b=20, \qquad c=11"))

display(Math(r"""
\alpha
=
1+4\left(\frac{c-a}{b-a}\right)
=
1+4\left(\frac{11-8}{20-8}\right)
"""))

display(Math(r"""
=
1+4\left(\frac{3}{12}\right)
=
2
"""))

display(Math(r"""
\beta
=
1+4\left(\frac{b-c}{b-a}\right)
=
1+4\left(\frac{20-11}{20-8}\right)
"""))

display(Math(r"""
=
1+4\left(\frac{9}{12}\right)
=
4
"""))

# Step 2
display(Markdown("## 2. Support"))

display(Math(r"8 \leq X \leq 20"))

# Step 3
display(Markdown("## 3. Probability Density Function"))

display(Math(r"""
f_X(x)=
\frac{
(x-8)^{\alpha-1}(20-x)^{\beta-1}
}{
B(\alpha,\beta)(20-8)^{\alpha+\beta-1}
},
\qquad
8\leq x\leq20
"""))

display(Markdown("Substituting $\\alpha=2$ and $\\beta=4$:"))

display(Math(r"""
f_X(x)=
\frac{
(x-8)(20-x)^3
}{
B(2,4)\,12^5
}
"""))

display(Math(r"""
B(2,4)
=
\frac{\Gamma(2)\Gamma(4)}{\Gamma(6)}
=
\frac{1!\,3!}{5!}
=
\frac{1}{20}
"""))

display(Math(r"""
f_X(x)=
\frac{
20(x-8)(20-x)^3
}{
12^5
},
\qquad 8\leq x\leq20
"""))

display(Math(r"f_X(x)=0 \qquad \text{otherwise}"))

# Step 4
display(Markdown("## 4. Probability Space"))

display(Math(r"""
\Omega=[8,20],
\qquad
\mathfrak{F}=\mathfrak{B}([8,20])
"""))

display(Math(r"""
P(A)=\int_A f_X(x)\,dx,
\qquad
A\in\mathfrak{F}
"""))

# Step 5
display(Markdown("## 5. Random Variable"))

display(Math(r"""
X(\omega)=\omega,
\qquad
\omega\in[8,20]
"""))

display(Markdown(
"Thus $X$ records the task duration in days."
))

# Step 6
display(Markdown("## 6. Expectation"))

display(Math(r"""
E[X]
=
\frac{a+4c+b}{6}
"""))

display(Math(r"""
E[X]
=
\frac{8+4(11)+20}{6}
"""))

display(Math(r"""
=
\frac{72}{6}
=
12
"""))

display(Math(r"E[X]=12"))

# Step 7
display(Markdown("## 7. Variance"))

display(Math(r"""
\mathrm{Var}(X)
=
\frac{(E[X]-a)(b-E[X])}{7}
"""))

display(Math(r"""
=
\frac{(12-8)(20-12)}{7}
"""))

display(Math(r"""
=
\frac{4\cdot8}{7}
=
\frac{32}{7}
\approx4.57
"""))

# Step 8
display(Markdown("## 8. Standard Deviation"))

display(Math(r"""
\sigma
=
\sqrt{\mathrm{Var}(X)}
=
\sqrt{\frac{32}{7}}
\approx2.14
"""))

display(Math(r"""
\sigma
\approx
\frac{b-a}{6}
=
\frac{20-8}{6}
=
2
"""))

# Step 9
display(Markdown("## 9. Comparison: Triangular vs PERT"))

display(Markdown(r"""
| Quantity | Triangular | PERT |
|---|---|---|
| Mean | $13$ | $12$ |
| Variance | $\dfrac{49}{6}\approx8.17$ | $\dfrac{32}{7}\approx4.57$ |
| Standard deviation | $\approx2.86$ | $\approx2.14$ |
| Main feature | Linear, sharp peak | Smooth, mode weighted more heavily |
"""))

display(Markdown(
"The PERT model pulls the expected value closer to the most likely case and produces a smaller variance than the Triangular model."
))

# Step 10
display(Markdown("## 10. Interpretation"))

display(Markdown("""
- $a=8$ is the shortest plausible completion time
- $b=20$ is the longest plausible completion time
- $c=11$ is the most likely completion time
- $E[X]=12$ is the expected task duration
- $\sigma\approx2.14$ measures the uncertainty in the task duration
"""))

display(Markdown(
"These values help project managers estimate schedules, assess deadline risks, and plan contingencies."
))

# ---------------------------------------------------
# GRAPH SECTION
# ---------------------------------------------------

display(Markdown("# Comparison Graph: Triangular vs PERT"))

# Parameters
a = 8
b = 20
c_mode = 11

# Triangular distribution parameters
c_scipy = (c_mode - a) / (b - a)
loc = a
scale = b - a

# PERT shape parameters
alpha_pert = 1 + 4 * (c_mode - a) / (b - a)
beta_pert = 1 + 4 * (b - c_mode) / (b - a)

# Generate points
x = np.linspace(a, b, 500)

# PDFs
y_triang = triang.pdf(x, c=c_scipy, loc=loc, scale=scale)
y_pert = beta.pdf(x, alpha_pert, beta_pert, loc=a, scale=b-a)

# Create figure
fig = go.Figure()

# Triangular graph
fig.add_trace(go.Scatter(
    x=x,
    y=y_triang,
    mode='lines',
    name='Triangular Distribution',
    line=dict(width=3, dash='dash')
))

# PERT graph
fig.add_trace(go.Scatter(
    x=x,
    y=y_pert,
    mode='lines',
    name='PERT Distribution',
    line=dict(width=3)
))

# Layout
fig.update_layout(
    title='Comparison: Triangular vs PERT Distribution',
    xaxis_title='Task Duration (days)',
    yaxis_title='Probability Density',
    template='plotly_white'
)

# Show graph
fig.show()
```

```python
from IPython.display import display, Math, Markdown

# ---------------------------------------------------
# TITLE
# ---------------------------------------------------

display(Markdown("# Symmetric PERT Distribution Derivation"))

# ---------------------------------------------------
# STEP 1
# ---------------------------------------------------

display(Markdown("## 1. PERT Mean Formula"))

display(Markdown("Use the PERT mean formula"))

display(Math(r"""
E[X]
=
\frac{a+4c+b}{6}
"""))

display(Markdown("---"))

# ---------------------------------------------------
# STEP 2
# ---------------------------------------------------

display(Markdown("## 2. Condition for Symmetry"))

display(Markdown(
"In general, if the expected value equals the most likely value, then"
))

display(Math(r"""
E[X]=c
"""))

display(Markdown("Substitute into the PERT mean formula:"))

display(Math(r"""
\frac{a+4c+b}{6}=c
"""))

display(Markdown("Multiply both sides by 6:"))

display(Math(r"""
a+4c+b=6c
"""))

display(Markdown("Rearrange terms:"))

display(Math(r"""
a+b=2c
"""))

display(Markdown("Hence"))

display(Math(r"""
b=2c-a
"""))

# ---------------------------------------------------
# STEP 3
# ---------------------------------------------------

display(Markdown("## 3. Numerical Example"))

display(Markdown("Thus"))

display(Math(r"""
b=17
"""))

display(Markdown("---"))

# ---------------------------------------------------
# STEP 4
# ---------------------------------------------------

display(Markdown("## 4. Variance Approximation"))

display(Markdown("Using the approximation"))

display(Math(r"""
\mathrm{Var}(X)
=
\frac{(E[X]-a)(b-E[X])}{7}
"""))

display(Markdown("and imposing"))

display(Math(r"""
E[X]=c
"""))

display(Markdown("we obtain"))

display(Math(r"""
\mathrm{Var}(X)
=
\frac{(c-a)(b-c)}{7}
"""))

# ---------------------------------------------------
# STEP 5
# ---------------------------------------------------

display(Markdown("## 5. Substitute the Symmetry Condition"))

display(Markdown("Since"))

display(Math(r"""
b=2c-a
"""))

display(Markdown("then"))

display(Math(r"""
b-c
=
(2c-a)-c
=
c-a
"""))

display(Markdown("Therefore"))

display(Math(r"""
\mathrm{Var}(X)
=
\frac{(c-a)^2}{7}
"""))

# ---------------------------------------------------
# STEP 6
# ---------------------------------------------------

display(Markdown("## 6. Equivalent Form"))

display(Markdown("Equivalently"))

display(Math(r"""
\mathrm{Var}(X)
=
\frac{(b-a)^2}{28}
"""))

# ---------------------------------------------------
# STEP 7
# ---------------------------------------------------

display(Markdown("## 7. Interpretation"))

display(Markdown(
"If the expected value equals the most likely value, then the PERT model becomes symmetric."
))

display(Math(r"""
\mathrm{Var}_{\text{PERT}}(X)
=
\frac{(c-a)^2}{7}
"""))

display(Markdown(
"This shows that the variability depends only on the spread between the optimistic estimate and the most likely estimate."
))
```
